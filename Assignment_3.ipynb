```python
import pandas as pd 
import numpy as np 
from scipy import stats
```

```python
df = pd.read_csv('Student_Performance_Dataset.csv')   # replace with dataset of choice 
print(df.head())
print(df.describe())
```

```python
mean_val = df['Age'].mean()
median_val = df['Age'].median() 
mode_val = df['Age'].mode()[0] 
print('Mean:', mean_val) 
print('Median:', median_val) 
print('Mode:', mode_val)
```

```python
marks_array = df['Age'].to_numpy()
print('NumPy Mean:', np.mean(marks_array))
print('NumPy Median:', np.median(marks_array))
print('SciPy Mode:', stats.mode(marks_array, keepdims=True))
```

```python
group_summary = df.groupby('Gender')['Age'].agg(['mean', 'median',    
lambda x: x.mode()[0]])
group_summary.columns = ['Mean', 'Median', 'Mode']
print(group_summary)
```

```python
if mean_val > median_val:
    print('Distribution is right (positively) skewed')
elif mean_val < median_val:
    print('Distribution is left (negatively) skewed')
else:
    print('Distribution is approximately symmetric')
```
