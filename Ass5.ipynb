```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm, skew, kurtosis
```

```python
data = np.array([
    45, 50, 52, 55, 56, 58, 60, 61, 62, 63,
    64, 65, 65, 66, 67, 68, 68, 69, 70, 70,
    71, 72, 72, 73, 74, 74, 75, 75, 76, 77,
    78, 78, 79, 80, 80, 81, 82, 83, 84, 85,
    86, 87, 88, 90, 92, 94, 95, 60, 70, 75
])

df = pd.DataFrame({'Score': data})

print("Student Test Scores Dataset:")
print(df)
```

```python
mean = np.mean(data)
std_dev = np.std(data)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std_dev, 2))
```

```python
x = np.linspace(
    mean - 4 * std_dev,
    mean + 4 * std_dev,
    1000
)

y = norm.pdf(x, mean, std_dev)

print("Normal distribution curve generated successfully.")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    data,
    bins=10,
    density=True,
    alpha=0.5,
    edgecolor='black',
    label='Actual Student Scores'
)

plt.plot(
    x, y,
    linewidth=2,
    label='Normal Distribution Curve'
)

plt.axvline(mean, linestyle='--', label='Mean')

plt.axvline(
    mean + std_dev,
    linestyle=':',
    label='+1 Standard Deviation'
)

plt.axvline(
    mean - std_dev,
    linestyle=':',
    label='-1 Standard Deviation'
)

plt.title('Normal Distribution of Student Test Scores')
plt.xlabel('Student Marks')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
within_1std = np.mean(
    (data >= mean - std_dev) &
    (data <= mean + std_dev)
) * 100

within_2std = np.mean(
    (data >= mean - 2 * std_dev) &
    (data <= mean + 2 * std_dev)
) * 100

within_3std = np.mean(
    (data >= mean - 3 * std_dev) &
    (data <= mean + 3 * std_dev)
) * 100

print("Empirical Rule Analysis:")
print(f"Within 1 standard deviation: {within_1std:.2f}%")
print(f"Within 2 standard deviations: {within_2std:.2f}%")
print(f"Within 3 standard deviations: {within_3std:.2f}%")
```

```python
print("Skewness:", round(skew(data), 3))
print("Kurtosis:", round(kurtosis(data), 3))
```
