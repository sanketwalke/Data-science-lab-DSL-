BUILT IN IMPLEMENTATION

```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("city_temperature.csv")
```

```python
# Select temperature data for two cities
city_a = df[df['City'] == 'Algiers']['AvgTemperature'].dropna()
city_b = df[df['City'] == 'Cairo']['AvgTemperature'].dropna()
```

```python
print("Algiers records:", len(city_a))
print("Cairo records:", len(city_b))
```

```python

# 1. Mean

mean_a = city_a.mean()
mean_b = city_b.mean()

print("Mean - Algiers:", mean_a)
print("Mean - Cairo:", mean_b)







```

```python
# 2. Range

range_a = city_a.max() - city_a.min()
range_b = city_b.max() - city_b.min()

print("\nRange - Algiers:", range_a)
print("Range - Cairo:", range_b)
```

```python
# 3. Variance
# Pandas .var() uses sample variance (ddof=1) by default

var_a = city_a.var()
var_b = city_b.var()

print("\nVariance - Algiers:", var_a)
print("Variance - Cairo:", var_b)
```

```python
# 4. Standard Deviation

std_a = city_a.std()
std_b = city_b.std()

print("\nStandard Deviation - Algiers:", std_a)
print("Standard Deviation - Cairo:", std_b)


```

```python
# 5. Coefficient of Variation

cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print("\nCV% - Algiers:", cv_a)
print("CV% - Cairo:", cv_b)
```

```python


if std_a < std_b:
    print("Algiers has more consistent temperatures than Cairo.")
else:
    print("Cairo has more consistent temperatures than Algiers.")
```

MANUALLY IMPLEMENTED

```python
import pandas as pd
import math

```

```python
# Load dataset
df = pd.read_csv("city_temperature.csv")

# Select temperature data for two cities
city_a = df[df['City'] == 'Algiers']['AvgTemperature'].dropna().tolist()
city_b = df[df['City'] == 'Cairo']['AvgTemperature'].dropna().tolist()
```

```python

print("Algiers records:", len(city_a))
print("Cairo records:", len(city_b))
```

```python
# 1. Mean

mean_a = sum(city_a) / len(city_a)
mean_b = sum(city_b) / len(city_b)

print("Mean - Algiers:", mean_a)
print("Mean - Cairo:", mean_b)


```

```python
# 2. Range
# Range = Maximum - Minimum

range_a = max(city_a) - min(city_a)
range_b = max(city_b) - min(city_b)

print("\nRange - Algiers:", range_a)
print("Range - Cairo:", range_b)


```

```python
# 3. Variance
# Sample Variance = Sum of squared deviations / (n - 1)

sum_squared_a = 0

for x in city_a:
    sum_squared_a += (x - mean_a) ** 2

variance_a = sum_squared_a / (len(city_a) - 1)


sum_squared_b = 0

for x in city_b:
    sum_squared_b += (x - mean_b) ** 2

variance_b = sum_squared_b / (len(city_b) - 1)

print("\nVariance - Algiers:", variance_a)
print("Variance - Cairo:", variance_b)


```

```python
# 4. Standard Deviation
# Standard Deviation = Square root of Variance

std_a = math.sqrt(variance_a)
std_b = math.sqrt(variance_b)

print("\nStandard Deviation - Algiers:", std_a)
print("Standard Deviation - Cairo:", std_b)


```

```python
# 5. Coefficient of Variation
# CV = (Standard Deviation / Mean) * 100

cv_a = (std_a / mean_a) * 100
cv_b = (std_b / mean_b) * 100

print("\nCV% - Algiers:", cv_a)
print("CV% - Cairo:", cv_b)
```

```python


if std_a < std_b:
    print("Algiers has more consistent temperatures than Cairo.")
else:
    print("Cairo has more consistent temperatures than Algiers.")
```
