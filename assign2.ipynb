```python
df["parking"] = df["parking"].fillna(df["parking"].mean())
```

```python
df["price_per_sqft"] = df["price"] / df["area"]
```

```python
mapping = {"yes":1, "no":0}

df["mainroad"] = df["mainroad"].map(mapping)
df["guestroom"] = df["guestroom"].map(mapping)
df["basement"] = df["basement"].map(mapping)
df["hotwaterheating"] = df["hotwaterheating"].map(mapping)
df["airconditioning"] = df["airconditioning"].map(mapping)
df["prefarea"] = df["prefarea"].map(mapping)
```

```python
df["bedrooms"] = df["bedrooms"].astype(int)
```

```python
df.rename(columns={"price":"HousePrice"}, inplace=True)
```

```python
df.sort_values(by="HousePrice", ascending=False)
```

```python
city = pd.DataFrame({
    "furnishingstatus":["furnished","semi-furnished","unfurnished"],
    "Category":["Premium","Standard","Budget"]
})

merged_df = pd.merge(df, city, on="furnishingstatus", how="left")

merged_df.head()
```
