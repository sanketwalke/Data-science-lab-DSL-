```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
n = 20       # Number of trials
p = 0.25     # Probability of success

print("Number of trials:", n)
print("Probability of success:", p)
print("Probability of failure:", 1 - p)
```

```python
x = np.arange(0, n + 1)

pmf = binom.pmf(x, n, p)
cdf = binom.cdf(x, n, p)

print("Number of Successes | PMF | CDF")

for i in range(len(x)):
    print(f"{x[i]:18} | {pmf[i]:.4f} | {cdf[i]:.4f}")
```

```python
mean, var = binom.stats(n, p, moments='mv')

std_dev = np.sqrt(var)

print("Mean (Expected Successes):", mean)
print("Variance:", var)
print("Standard Deviation:", std_dev)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    x, pmf,
    edgecolor='black',
    label='PMF'
)

plt.axvline(
    mean,
    linestyle='--',
    label='Mean'
)

plt.title(f'Binomial Distribution (n={n}, p={p})')
plt.xlabel('Number of Successes')
plt.ylabel('Probability')
plt.xticks(x)
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
p_at_least_5 = 1 - binom.cdf(4, n, p)

print("P(X >= 5):", round(p_at_least_5, 4))
print("Probability in percentage:", round(p_at_least_5 * 100, 2), "%")
```

```python
p_exactly_8 = binom.pmf(8, n, p)

print("P(X = 8):", round(p_exactly_8, 4))
print("Probability in percentage:", round(p_exactly_8 * 100, 2), "%")
```

```python
p_at_most_5 = binom.cdf(5, n, p)

print("P(X <= 5):", round(p_at_most_5, 4))
print("Probability in percentage:", round(p_at_most_5 * 100, 2), "%")
```

```python
plt.figure(figsize=(8, 5))

plt.step(x, cdf, where='mid', label='CDF')
plt.scatter(x, cdf)

plt.title('Cumulative Binomial Distribution')
plt.xlabel('Number of Successes')
plt.ylabel('Cumulative Probability')
plt.xticks(x)
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
